# Miletos Configuration Catalog

Inspect and run the complete catalog of target-analysis configurations. The catalog spans transiting planets, stellar variability, flares, compact objects, survey populations, and simulated observations. Many observational configurations require network access or local input products.

```python
from IPython.display import Image, display

from examples.catalog import run as catalog
from miletos.paths import get_repository_path
```

## List Available Configurations

Each entry maps to one configuration function in the maintained catalog.

```python
configurations = {
    name: value
    for name, value in vars(catalog).items()
    if name.startswith("cnfg_") and callable(value)
}
print(f"Available configurations: {len(configurations)}")
for name in sorted(configurations):
    print(name.removeprefix("cnfg_"))
```

## Run a Configuration

The deterministic simulated-transit configuration is the offline default. Replace the selected name to run another catalog workflow and provide any required positional arguments documented by that function.

```python
configuration_name = "cnfg_simulated_transit_diagnostic"
configurations[configuration_name](typefileplot="png")

output_path = (
    get_repository_path()
    / "examples"
    / "catalog"
    / "visuals"
    / "example_catalog_transit.png"
)
print(f"Reading from {output_path}...")
display(Image(filename=str(output_path)))

assert output_path.is_file()
```
