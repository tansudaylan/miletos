# JWST ERS Analysis of WASP-39b

Reproduce the public Near Infrared Spectrograph G395H analysis from Alderson et al. (2023), DOI `10.1038/s41586-022-05591-3`. Miletos downloads and caches the Zenodo products, verifies the archived reduction identities, compares the transmission spectrum with the published ATMO model, and writes six figures.

```python
from IPython.display import Image, display

from miletos.ers import run_wasp39_ers_g395h_reproduction
from miletos.paths import get_repository_path
```

## Run Miletos

The first run requires network access. Later runs use the cached public archive unless `refresh_data=True` is requested.

```python
result = run_wasp39_ers_g395h_reproduction(
    typefileplot="png",
    refresh_data=False,
)
visual_path = get_repository_path() / "examples" / "WASP-39b" / "visuals"
figure_paths = sorted(visual_path.glob("wasp39_ers_g395h_*.png"))
```

## Inspect the Products

Summarize the published wavelength bins and display every Miletos-generated figure.

```python
print(f"Published wavelength bins: {result.wavelength_microns.size}")
print(f"Reduced chi-squared: {result.reduced_chi_squared:.2f}")
for figure_path in figure_paths:
    print(f"Reading from {figure_path}...")
    display(Image(filename=str(figure_path)))

assert len(figure_paths) == 6
```
