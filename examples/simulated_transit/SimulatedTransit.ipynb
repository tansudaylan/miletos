# Simulated Transit Pipeline

Run the deterministic Miletos transit benchmark. Miletos generates the simulated light curve, masks the known transits during spline detrending, phase-folds the result, propagates the uncertainties, and writes the focused pipeline figure.

```python
from contextlib import redirect_stdout
from io import StringIO

from IPython.display import Image, display

from miletos.diagnostics import run_simulated_transit_diagnostic
from miletos.paths import get_repository_path
```

## Run Miletos

The reusable API owns the complete calculation and figure generation.

```python
output_path = (
    get_repository_path()
    / "examples"
    / "simulated_transit"
    / "visuals"
    / "simulated_transit_diagnostic.png"
)
with redirect_stdout(StringIO()):
    result = run_simulated_transit_diagnostic(output_path)
```

## Inspect the Result

Report the recovered transit depth and display the Miletos-generated figure.

```python
print(f"Orbital period: {result.period_days:.2f} [day]")
display(Image(filename=str(output_path)))
```
